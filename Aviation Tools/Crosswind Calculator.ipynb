```python
import math, numpy as np

rwy_hdg = int(input('Runway heading: '))
wind_hdg = int(input('Wind heading: '))
wind_v = int(input('Wind Speed: '))


rwy_hdg_rad = math.radians(rwy_hdg)
wind_hdg_rad = math.radians(wind_hdg)
```

```python
relative_angle = abs(wind_hdg_rad - rwy_hdg_rad)

def split_component(magnitude, angle):
    headwind_v = round(magnitude * math.cos(angle), 1)
    crosswind_v = round(magnitude * math.sin(angle), 1)

    return headwind_v, crosswind_v

headwind_speed, crosswind_speed = split_component(wind_v, relative_angle)

print(f'Headwind Speed: {headwind_speed}, Crosswind Speed: {crosswind_speed}')

if headwind_speed < 0:
    print('Tailwind present; use alternate runway if able')
```
