```python
import numpy as np, math, matplotlib.pyplot as plt
```

```python
#Initialize Vars
g = -9.8
vel_0 = 25 
ang_initial = 30 
y_initial = 5 

ang_rad = math.radians(ang_initial) 

#Array Setup
time = np.linspace(0 , 20, 1000)
velx_0 = vel_0 * math.cos(ang_rad)
vely_0 = vel_0 * math.sin(ang_rad)


def parametric(parameter, position, velocity, acceleration):
    position = (0.5 * acceleration * parameter ** 2) + (velocity * parameter) + position
    return position

x_coords = parametric(time, 0, velx_0, 0)
y_coords = parametric(time, y_initial, vely_0, g)
y_coords = y_coords

print(x_coords[0])
print(time[0])
```

```python
#Solutions
def solve(t, x, y):
    ground_intersect = y <= 0 
    impact = t[ground_intersect][0]
    delta_x = x[ground_intersect][0]
    peak = y[np.argmax(y)]
    return impact, delta_x, peak

impact_time, displacement,  y_max = solve(time, x_coords, y_coords)
```

```python
#Plot Data

def title_graph(title, x_label, y_label):
    plt.title(title)
    plt.xlabel(x_label)
    plt.ylabel(y_label)

def draw_graph(title, x_label, y_label, x, y):
   
    plt.figure()
    title_graph(title, x_label, y_label)
    plt.plot(x, y)
   
    #Limits !!!! USING GLOBAL VARIABLES; FIX TO PROPERLY SCALE ALL DATASETS !!!!
    y_peak = y_coords[np.argmax(y_coords)]
    plt.ylim(0, y_max + .075 * y_max )
    plt.xlim(0, displacement + 0.05 * displacement) 

#Draw Graphs
draw_graph('Height v. Displacement', 'Displacement (m)', 'Height (m)', x_coords, y_coords)
draw_graph('Height v. Time', 'Time (s)', 'Height (m)', time, y_coords)
draw_graph('Displacement v. Time', 'Time (s)', 'Displacement (m)', time, x_coords)
```
